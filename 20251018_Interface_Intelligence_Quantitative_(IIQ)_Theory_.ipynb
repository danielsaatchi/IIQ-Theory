```python
# ============================================================
#  Interface Intelligence Quantified (IIQ) Simulation
#  Colab-Compatible Version
#  Developed by DANIEL SAATCHI, Computational Scientist & AI Developer
#  Date: 2025 October 18th
# ============================================================

# --- Mount Google Drive ---
from google.colab import drive
drive.mount('/content/drive')

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# --- Define output directory ---
output_dir = "/content/drive/MyDrive/IIQ-Scenarios"
os.makedirs(output_dir, exist_ok=True)
print(f"All files will be saved to: {output_dir}")

# ------------------------------------------------------------
# Define scenario dictionary (editable parameters) for Emerging Intelligences
# ------------------------------------------------------------
scenarios = {
    "Human-SFID AI System": {
        "β_b": 1, "C_b": 1,
        "β_m": 1, "C_m": 1,
        "Description": "Human AI interface with balanced machine and biological bandwidths."
    },
    "Orca/Dophine SFID AI System": {
        "β_b": 2, "C_b": 2,
        "β_m": 0, "C_m": 0,
        "Description": "Orca or dolphine operator and SFID device with moderate biological and machine bandwidth."
    },
    "Bat–SFID Tag": {
        "β_b": 0, "C_b": 2,
        "β_m": 0, "C_m": 2,
        "Description": "Bat with sonar capability interacting with a tagged AI classifier."
    },
    "Bornean Plant–SFID Classifier": {
        "β_b": 0, "C_b": 1,
        "β_m": 1, "C_m": 1,
        "Description": "Plant with passive biological input connected to high-capacity AI classifier."
    },
    "Bat/Dolphin/Orca + SFID AI System 2.0": {
        "β_b": 2, "C_b": 2,
        "β_m": 1, "C_m": 1,
        "Description": "Marine/air mammals interacting with SFID 2.0 AI system."
    },
    "Bat↔Human through SFID AI System 3.0": {
        "β_b": 2, "C_b": 3,
        "β_m": 3, "C_m": 2,
        "Description": "Advanced triadic interface among human, bat, and SFID 3.0 AI system."
    },
}

# ------------------------------------------------------------
# Convert to DataFrame
# ------------------------------------------------------------
df = pd.DataFrame.from_dict(scenarios, orient="index").reset_index()
df.rename(columns={"index": "System"}, inplace=True)

# ------------------------------------------------------------
# Compute Interface Intelligence (I = β_b * C_m + β_m * C_b)
# ------------------------------------------------------------
df["I"] = df["β_b"] * df["C_m"] + df["β_m"] * df["C_b"]

# ------------------------------------------------------------
# Interpret Interconsciousness based on Computed I Scoring
# ------------------------------------------------------------
def interpret(i):
    if i >= 4:
        return "High Interconscious (Multi Bio-AI Agents)"
    elif i >= 2:
        return "Moderate Interconsciousness (Bio-AI Cybernetics)"
    elif i >= 1:
        return "Low Interface Awareness (Debated)"
    else:
        return "Non-Interconscious"

df["Interconscious_Level"] = df["I"].apply(interpret)

# ------------------------------------------------------------
# Save results to Drive
# ------------------------------------------------------------
csv_path = f"{output_dir}/IIQ_scenarios_results.csv"
df.to_csv(csv_path, index=False)
print(f"Results saved to: {csv_path}")

# ------------------------------------------------------------
# Visualization 1: Bar Chart
# ------------------------------------------------------------
plt.figure(figsize=(10, 5))
plt.bar(df["System"], df["I"], color="teal")
plt.title("Interface Intelligence (I) Across IIQ Scenarios")
plt.ylabel("Interface Intelligence Score (I)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
bar_chart_path = f"{output_dir}/IIQ_bar_chart.png"
plt.savefig(bar_chart_path)
plt.close()

# ------------------------------------------------------------
# Visualization 2: Scatter Plot of Cross-Terms
# ------------------------------------------------------------
cross_terms = [(row["β_b"] * row["C_m"], row["β_m"] * row["C_b"]) for _, row in df.iterrows()]
x, y = zip(*cross_terms)
plt.figure(figsize=(6, 6))
plt.scatter(x, y, s=[v * 25 for v in df["I"]], alpha=0.8, c=df["I"], cmap="viridis", edgecolors="black")
plt.title("Cross-Terms of Interface Intelligence")
plt.xlabel("β_b × C_m (Bio → Machine)")
plt.ylabel("β_m × C_b (Machine → Bio)")
plt.tight_layout()
scatter_chart_path = f"{output_dir}/IIQ_scatter_plot.png"
plt.savefig(scatter_chart_path)
plt.close()

# ------------------------------------------------------------
# Visualization 3: Heatmap of Parameter Sweep
# ------------------------------------------------------------
beta_b = np.linspace(0, 3, 30)
C_b = np.linspace(0, 3, 30)
B, C = np.meshgrid(beta_b, C_b)
I = B * 2 + 2 * C  # Example: assuming β_m=2, C_m=2
plt.figure(figsize=(6, 5))
plt.imshow(I, origin="lower", extent=[0, 3, 0, 3], aspect="auto", cmap="plasma")
plt.colorbar(label="I Value")
plt.xlabel("β_b (Biological Bandwidth)")
plt.ylabel("C_b (Biological Cognition)")
plt.title("IIQ Parameter Sweep Heatmap")
plt.tight_layout()
heatmap_path = f"{output_dir}/IIQ_heatmap.png"
plt.savefig(heatmap_path)
plt.close()

# ------------------------------------------------------------
# Save summary text file
# ------------------------------------------------------------
readme_path = f"{output_dir}/README.txt"
with open(readme_path, "w") as f:
    f.write(
        "IIQ (Interface Intelligence Quantified) Simulation Results\n"
        "=========================================================\n"
        "Folder contents:\n"
        "• IIQ_scenarios_results.csv – All computed data.\n"
        "• IIQ_bar_chart.png – Intelligence comparison.\n"
        "• IIQ_scatter_plot.png – Cross-term visualization.\n"
        "• IIQ_heatmap.png – Parameter sweep map.\n"
        "• README.txt – This summary.\n\n"
        "Adjust β_b, C_b, β_m, and C_m values in the dictionary above to test new systems.\n"
    )

print("\n✅ Simulation complete!")
print(f"Files saved to: {output_dir}")
```

```python
# ============================================================
#  Corporate Interface Intelligence (CII) Simulation (Full)
#  Colab-ready, journal-style visuals
#  Developed by Daniel Saatchi — Computational Scientist & AI Developer
#  Date: 2025 October 22
#  FIXED: Annotation positioning in all scatter plots
# ============================================================

# --- Mount Google Drive (works in Colab) ---
try:
    from google.colab import drive
    drive.mount('/content/drive', force_remount=False)
    BASE_DIR = "/content/drive/MyDrive/CII-Scenarios"
except Exception:
    # If not running in Colab, fall back to local folder
    BASE_DIR = "./CII-Scenarios"
import os
os.makedirs(BASE_DIR, exist_ok=True)
print("Output folder:", BASE_DIR)

# --- Imports ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from scipy.cluster.hierarchy import fclusterdata

# --- Journal-style Matplotlib config ---
mpl.rcParams.update({
    "font.size": 12,
    "font.family": "serif",
    "axes.labelweight": "bold",
    "axes.titlesize": 14,
    "axes.titlepad": 10,
    "figure.dpi": 200
})

# ------------------------------------------------------------
# Scenario dictionary (editable)
# ------------------------------------------------------------
scenarios = {
    "Google": {"Cb":183323, "Bb":2, "Bm":7, "GPUs":400000, "LLM_score":89/100, "TokenCost":0.003},
    "DeepMind": {"Cb":6724, "Bb":2, "Bm":7, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Elon Musk (Neuralink–xAI)": {"Cb":14500, "Bb":3, "Bm":9, "GPUs":400000, "LLM_score":86/100, "TokenCost":0.003},
    "OpenAI": {"Cb":3000, "Bb":2, "Bm":7, "GPUs":400000, "LLM_score":88/100, "TokenCost":0.003},
    "Anthropic": {"Cb":1300, "Bb":2, "Bm":7, "GPUs":400000, "LLM_score":92/100, "TokenCost":0.003},
    "Microsoft AI": {"Cb":12000, "Bb":2, "Bm":7, "GPUs":400000, "LLM_score":88/100, "TokenCost":0.003},
    "Apple": {"Cb":164000, "Bb":2, "Bm":8, "GPUs":400000, "LLM_score":70/100, "TokenCost":0.003},
    "Meta": {"Cb":75945, "Bb":2, "Bm":8, "GPUs":400000, "LLM_score":76/100, "TokenCost":0.003},
    "Transhuman Corp (Extreme Emerge 1)": {"Cb":1300, "Bb":4, "Bm":9, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Alien Corp (Extreme Emerge 2)": {"Cb":1300, "Bb":5, "Bm":5, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Transhuman Startup (Extreme Emerge 3)": {"Cb":2, "Bb":5, "Bm":7, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Alien Transpecies (Extreme Emerge 4)": {"Cb":1, "Bb":5, "Bm":5, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Alien Transpecies (Extreme Emerge 5)": {"Cb":1, "Bb":6, "Bm":5, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Alien Transpecies (Extreme Emerge 6)": {"Cb":1, "Bb":6, "Bm":9, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Alien Transpecies (Extreme Emerge 7)": {"Cb":1, "Bb":9, "Bm":9, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Super Intelligences (Extreme Emerge 8)": {"Cb":0, "Bb":9, "Bm":9, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Super Intelligences (Extreme Emerge 9)": {"Cb":0, "Bb":9, "Bm":1000, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Super Intelligences (Extreme Emerge 10)": {"Cb":0, "Bb":2, "Bm":5000, "GPUs":400000, "LLM_score":99/100, "TokenCost":0.003},
  #  "Super Intelligences (Extreme Emerge 11)": {"Cb":0, "Bb":9, "Bm":5000, "GPUs":400000, "LLM_score":99/100, "TokenCost":0.003},
    "Alien Transpecies (Extreme Emerge 12)": {"Cb":1, "Bb":9, "Bm":5000, "GPUs":400000, "LLM_score":99/100, "TokenCost":0.003},
}

# ------------------------------------------------------------
# Convert to DataFrame
# ------------------------------------------------------------
df = pd.DataFrame.from_dict(scenarios, orient='index').reset_index().rename(columns={'index':'Entity'})

# ------------------------------------------------------------
# Compute metrics
# ------------------------------------------------------------
df['Cm'] = df['GPUs'] * df['LLM_score']
df['Bio_to_Machine'] = df['Bb'] * df['Cm']
df['Machine_to_Bio'] = df['Bm'] * df['Cb']
df['I'] = df['Bio_to_Machine'] + df['Machine_to_Bio']

df['Denom'] = df['GPUs'] + df['Cb']
df['Denom'] = df['Denom'].replace(0, np.nan)
df['VPT'] = df['TokenCost'] * df['I'] / df['Denom']
df['TFR'] = df['VPT'] / df['TokenCost']

df['VPT'] = df['VPT'].fillna(0)
df['TFR'] = df['TFR'].fillna(0)

# Save numeric results
csv_path = os.path.join(BASE_DIR, "CII_scenarios_results.csv")
df.to_csv(csv_path, index=False)
print("Saved CSV:", csv_path)

print(df[['Entity','Cb','Bb','Bm','GPUs','LLM_score','Cm','Bio_to_Machine','Machine_to_Bio','I','VPT','TFR']].head(6))

# ------------------------------------------------------------
# Helper functions for annotation
# ------------------------------------------------------------
def estimate_label_box(label_x, label_y, text_length, ha, x_range, y_range):
    """Estimate the bounding box of a label in data coordinates"""
    # More conservative estimate accounting for actual rendered text
    char_width = x_range * 0.006  # Increased from 0.004
    label_width = char_width * text_length
    label_height = y_range * 0.035  # Increased from 0.025 for safety

    if ha == 'right':
        x_left = label_x - label_width
        x_right = label_x + x_range * 0.01  # Small padding
    else:
        x_left = label_x - x_range * 0.01  # Small padding
        x_right = label_x + label_width

    y_bottom = label_y - label_height/2
    y_top = label_y + label_height/2

    return (x_left, x_right, y_bottom, y_top)

def boxes_overlap(box1, box2, buffer, x_range, y_range):
    """Check if two bounding boxes overlap with a buffer"""
    x1_left, x1_right, y1_bottom, y1_top = box1
    x2_left, x2_right, y2_bottom, y2_top = box2

    buffer_x = x_range * buffer
    buffer_y = y_range * buffer

    if (x1_right + buffer_x < x2_left or x2_right + buffer_x < x1_left or
        y1_top + buffer_y < y2_bottom or y2_top + buffer_y < y1_bottom):
        return False
    return True

def add_smart_annotations(ax, x_data, y_data, labels, x_range, y_range, x_min, x_max, y_min, y_max):
    """Add annotations with collision avoidance"""
    # More conservative margins
    margin_x = x_range * 0.08
    margin_y = y_range * 0.08

    coords = np.column_stack([x_data.values, y_data.values])

    try:
        clusters = fclusterdata(coords, t=50, criterion='distance', metric='euclidean')
    except:
        clusters = np.arange(len(coords))

    df_sorted = pd.DataFrame({
        'x_plot': x_data.values,
        'y_plot': y_data.values,
        'label': labels,
        'cluster': clusters
    })
    df_sorted = df_sorted.sort_values(['cluster', 'x_plot', 'y_plot'])

    used_label_boxes = []

    for idx, row in df_sorted.iterrows():
        x_pos = row['x_plot']
        y_pos = row['y_plot']
        txt = row['label']

        # Determine base positioning
        if x_pos > x_data.median():
            ha = 'right'
            base_offset_x = -15
            if x_pos > x_max - margin_x * 1.5:
                base_offset_x = -35
        else:
            ha = 'left'
            base_offset_x = 15
            if x_pos < x_min + margin_x * 1.5:
                base_offset_x = 35

        best_offset_y = 0
        best_score = -1
        best_box = None

        # Try vertical positions
        for try_offset in range(-120, 121, 6):
            label_x = x_pos + base_offset_x * x_range / 1000
            label_y = y_pos + try_offset * y_range / 700

            label_box = estimate_label_box(label_x, label_y, len(txt), ha, x_range, y_range)
            x_left, x_right, y_bottom, y_top = label_box

            # Stricter boundary checking with margins
            if (y_bottom < y_min + margin_y * 0.3 or
                y_top > y_max - margin_y * 0.3 or
                x_left < x_min + margin_x * 0.3 or
                x_right > x_max - margin_x * 0.3):
                continue

            has_overlap = False
            min_dist = float('inf')

            for existing_box in used_label_boxes:
                if boxes_overlap(label_box, existing_box, 0.02, x_range, y_range):
                    has_overlap = True
                    break

                exist_center_x = (existing_box[0] + existing_box[1]) / 2
                exist_center_y = (existing_box[2] + existing_box[3]) / 2
                label_center_x = (x_left + x_right) / 2
                label_center_y = (y_bottom + y_top) / 2
                dist = np.sqrt((exist_center_x - label_center_x)**2 +
                              (exist_center_y - label_center_y)**2)
                min_dist = min(min_dist, dist)

            if has_overlap:
                continue

            score = min_dist - abs(try_offset) * y_range / 10000

            if score > best_score:
                best_score = score
                best_offset_y = try_offset
                best_box = label_box

                if min_dist > x_range / 15 and abs(try_offset) < 40:
                    break

        if best_box is not None:
            used_label_boxes.append(best_box)

        ax.annotate(txt, (x_pos, y_pos),
                     xytext=(base_offset_x, best_offset_y),
                     textcoords='offset points',
                     fontsize=6,
                     weight='normal',
                     ha=ha,
                     va='center',
                     arrowprops=dict(arrowstyle='-', color='gray', lw=0.4, alpha=0.7),
                     bbox=dict(boxstyle='round,pad=0.2', facecolor='white',
                              edgecolor='none', alpha=0.75))

    # Set limits with margins
    ax.set_xlim(x_min - margin_x, x_max + margin_x)
    ax.set_ylim(y_min - margin_y, y_max + margin_y)

# ------------------------------------------------------------
# Visualization A: Bar chart of I (scaled)
# ------------------------------------------------------------
scale = 1e6

plt.figure(figsize=(11,6))
plt.bar(df['Entity'], df['I']/scale, color='darkcyan', edgecolor='black', linewidth=0.6)
plt.title('Corporate Interface Intelligence (CII) — Scaled', weight='bold')
plt.ylabel(f'II (×{int(scale):,})', weight='bold')
plt.xticks(rotation=60, ha='right', fontsize=9)
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.tight_layout()
bar_out = os.path.join(BASE_DIR, "CII_I_Bar_Scaled.png")
plt.savefig(bar_out, bbox_inches='tight', dpi=200)
plt.show()
plt.close()
print("Saved bar chart:", bar_out)

# ------------------------------------------------------------
# Visualization B: Cross-term scatter (FIXED ANNOTATIONS)
# ------------------------------------------------------------
x = df['Bio_to_Machine'] / scale
y = df['Machine_to_Bio'] / scale
sizes = (df['I'] / df['I'].max()) * 800 + 50
c = df['I']

fig, ax = plt.subplots(figsize=(12,7))
sc = ax.scatter(x, y, s=sizes, c=c, cmap='viridis', edgecolors='k', alpha=0.9)
ax.set_xlabel('β_b × C_m (Bio → Machine) (scaled)', weight='bold')
ax.set_ylabel('β_m × C_b (Machine → Bio) (scaled)', weight='bold')
ax.set_title('CII Cross-Terms: Interhumanism Realm', weight='bold', pad=12)
ax.grid(linestyle=':', alpha=0.5)
cbar = plt.colorbar(sc, ax=ax)
cbar.set_label('Interface Intelligence (absolute)', weight='bold')

x_min, x_max = x.min(), x.max()
y_min, y_max = y.min(), y.max()
x_range = x_max - x_min
y_range = y_max - y_min

add_smart_annotations(ax, x, y, df['Entity'].values, x_range, y_range, x_min, x_max, y_min, y_max)

plt.tight_layout()
cross_out = os.path.join(BASE_DIR, "CII_CrossTerm_Scatter.png")
plt.savefig(cross_out, bbox_inches='tight', dpi=200)
plt.show()
plt.close()
print("Saved cross-term scatter:", cross_out)

# ------------------------------------------------------------
# Visualization C: Cross-term vs I
# ------------------------------------------------------------
plt.figure(figsize=(10,6))
plt.scatter(df['Bio_to_Machine']/scale, df['I']/scale, s=100, c='tab:blue', label='Bio→Machine (β_b×Cm)')
plt.scatter(df['Machine_to_Bio']/scale, df['I']/scale, s=100, c='tab:orange', label='Machine→Bio (β_m×Cb)')
plt.xlabel('Cross-term (scaled)', weight='bold')
plt.ylabel('Interface Intelligence (scaled)', weight='bold')
plt.title('Cross-term contribution to II (comparison)', weight='bold')
plt.legend()
plt.grid(linestyle=':', alpha=0.5)
cross_vs_I_out = os.path.join(BASE_DIR, "CII_Crossterm_vs_I.png")
plt.savefig(cross_vs_I_out, bbox_inches='tight', dpi=200)
plt.show()
plt.close()
print("Saved cross-term vs I plot:", cross_vs_I_out)

# ------------------------------------------------------------
# Visualization D: Economic scatter I vs VPT (FIXED ANNOTATIONS)
# ------------------------------------------------------------
x_econ = df['I'] / scale
y_econ = df['VPT']

fig2, ax2 = plt.subplots(figsize=(11,6))
sc2 = ax2.scatter(x_econ, y_econ, s=(df['TFR']/df['TFR'].max())*800+50,
                  c=df['TFR'], cmap='plasma', edgecolors='k', alpha=0.9)
ax2.set_xlabel('Interface Intelligence (scaled)', weight='bold')
ax2.set_ylabel('Value Per Token (VPT) [$]', weight='bold')
ax2.set_title('Economic Efficiency: VPT vs Interface Intelligence (II)', weight='bold')
cbar2 = plt.colorbar(sc2, ax=ax2)
cbar2.set_label('Token Financial Ratio (TFR)', weight='bold')
ax2.grid(linestyle=':', alpha=0.5)

x_min_e, x_max_e = x_econ.min(), x_econ.max()
y_min_e, y_max_e = y_econ.min(), y_econ.max()
x_range_e = x_max_e - x_min_e
y_range_e = y_max_e - y_min_e

add_smart_annotations(ax2, x_econ, y_econ, df['Entity'].values, x_range_e, y_range_e, x_min_e, x_max_e, y_min_e, y_max_e)

plt.tight_layout()
econ_out = os.path.join(BASE_DIR, "CII_Economic_VPT_vs_II.png")
plt.savefig(econ_out, bbox_inches='tight', dpi=200)
plt.show()
plt.close()
print("Saved economic scatter:", econ_out)

# ------------------------------------------------------------
# Visualization E: Heatmap Parameter Sweep
# ------------------------------------------------------------
beta_m_fixed = int(df['Bm'].median())
gpu_fixed = int(df['GPUs'].median())
Bb_range = np.linspace(0, 10, 200)
Cb_range = np.linspace(0, df['Cb'].max(), 200)
BB, CC = np.meshgrid(Bb_range, Cb_range)
I_grid = BB * (gpu_fixed * df['LLM_score'].median()) + beta_m_fixed * CC

plt.figure(figsize=(9,6))
im = plt.imshow(I_grid / scale, origin='lower', aspect='auto', cmap='inferno',
                extent=[Bb_range.min(), Bb_range.max(), Cb_range.min(), Cb_range.max()])
plt.colorbar(im, label=f'I (scaled by {int(scale):,})')
plt.xlabel('β_b (Biological Bandwidth)', weight='bold')
plt.ylabel('C_b (Biological Cognitive Capacity)', weight='bold')
plt.title('Parameter Sweep: II over β_b and C_b (machine fixed)', weight='bold')
plt.tight_layout()
heat_out = os.path.join(BASE_DIR, "CII_Parameter_Sweep.png")
plt.savefig(heat_out, bbox_inches='tight', dpi=200)
plt.show()
plt.close()
print("Saved parameter sweep heatmap:", heat_out)

# ------------------------------------------------------------
# Final save
# ------------------------------------------------------------
final_csv = os.path.join(BASE_DIR, "CII_scenarios_results_full.csv")
df.to_csv(final_csv, index=False)
print("Final CSV saved:", final_csv)

print("\n✅ All plots and CSVs saved to folder:", BASE_DIR)
```

```python
# ============================================================
#  Corporate Interface Intelligence (CII) - Economic VPT Scatter Animation
#  Animated scatter plot showing I vs VPT as GPUs vary
#  Developed by Daniel Saatchi — Computational Scientist & AI Developer
#  Date: 2025 October 24
# ============================================================

# --- Mount Google Drive (works in Colab) ---
try:
    from google.colab import drive
    drive.mount('/content/drive', force_remount=False)
    BASE_DIR = "/content/drive/MyDrive/CII-Scenarios"
except Exception:
    BASE_DIR = "./CII-Scenarios"
import os
os.makedirs(BASE_DIR, exist_ok=True)
print("Output folder:", BASE_DIR)

# --- Imports ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import HTML

# --- Journal-style Matplotlib config ---
mpl.rcParams.update({
    "font.size": 10,
    "font.family": "serif",
    "axes.labelweight": "bold",
    "axes.titlesize": 14,
    "axes.titlepad": 10,
    "figure.dpi": 100
})

# ------------------------------------------------------------
# CONFIGURATION: Change these parameters as needed
# ------------------------------------------------------------
PARAM_TO_VARY = "LLM_score"  # Parameter to animate
PARAM_MIN = 0.50             # Starting value (50% score)
PARAM_MAX = 0.99             # Ending value (99% score)
N_FRAMES = 120               # Number of animation frames

# ------------------------------------------------------------
# Scenario dictionary (base values)
# ------------------------------------------------------------
scenarios = {
    "Google": {"Cb":183323, "Bb":2, "Bm":7, "GPUs":400000, "LLM_score":89/100, "TokenCost":0.003},
    "DeepMind": {"Cb":6724, "Bb":2, "Bm":7, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Elon Musk (Neuralink–xAI)": {"Cb":14500, "Bb":3, "Bm":9, "GPUs":400000, "LLM_score":86/100, "TokenCost":0.003},
    "OpenAI": {"Cb":3000, "Bb":2, "Bm":7, "GPUs":400000, "LLM_score":88/100, "TokenCost":0.003},
    "Anthropic": {"Cb":1300, "Bb":2, "Bm":7, "GPUs":400000, "LLM_score":92/100, "TokenCost":0.003},
    "Microsoft AI": {"Cb":12000, "Bb":2, "Bm":7, "GPUs":400000, "LLM_score":88/100, "TokenCost":0.003},
    "Apple": {"Cb":164000, "Bb":2, "Bm":8, "GPUs":400000, "LLM_score":70/100, "TokenCost":0.003},
    "Meta": {"Cb":75945, "Bb":2, "Bm":8, "GPUs":400000, "LLM_score":76/100, "TokenCost":0.003},
    "Transhuman Corp (Extreme Emerge 1)": {"Cb":1300, "Bb":4, "Bm":9, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Alien Corp (Extreme Emerge 2)": {"Cb":1300, "Bb":5, "Bm":5, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Transhuman Startup (Extreme Emerge 3)": {"Cb":2, "Bb":5, "Bm":7, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Alien Transpecies (Extreme Emerge 4)": {"Cb":1, "Bb":5, "Bm":5, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Alien Transpecies (Extreme Emerge 5)": {"Cb":1, "Bb":6, "Bm":5, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Alien Transpecies (Extreme Emerge 6)": {"Cb":1, "Bb":6, "Bm":9, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Alien Transpecies (Extreme Emerge 7)": {"Cb":1, "Bb":9, "Bm":9, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Super Intelligences (Extreme Emerge 8)": {"Cb":0, "Bb":9, "Bm":9, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Super Intelligences (Extreme Emerge 9)": {"Cb":0, "Bb":9, "Bm":1000, "GPUs":400000, "LLM_score":90/100, "TokenCost":0.003},
    "Super Intelligences (Extreme Emerge 10)": {"Cb":0, "Bb":2, "Bm":5000, "GPUs":400000, "LLM_score":99/100, "TokenCost":0.003},
    "Alien Transpecies (Extreme Emerge 12)": {"Cb":1, "Bb":9, "Bm":5000, "GPUs":400000, "LLM_score":99/100, "TokenCost":0.003},
}

# ------------------------------------------------------------
# Function to calculate metrics for given parameter value
# ------------------------------------------------------------
def calculate_metrics(df, param_value):
    """Calculate all metrics including VPT and TFR for a given parameter value"""
    df_calc = df.copy()

    if PARAM_TO_VARY == "GPUs":
        df_calc['GPUs'] = param_value
    elif PARAM_TO_VARY == "LLM_score":
        df_calc['LLM_score'] = param_value
    elif PARAM_TO_VARY == "Bb":
        df_calc['Bb'] = param_value
    elif PARAM_TO_VARY == "Bm":
        df_calc['Bm'] = param_value
    elif PARAM_TO_VARY == "Cb":
        df_calc['Cb'] = param_value

    df_calc['Cm'] = df_calc['GPUs'] * df_calc['LLM_score']
    df_calc['Bio_to_Machine'] = df_calc['Bb'] * df_calc['Cm']
    df_calc['Machine_to_Bio'] = df_calc['Bm'] * df_calc['Cb']
    df_calc['I'] = df_calc['Bio_to_Machine'] + df_calc['Machine_to_Bio']

    # Calculate economic metrics
    df_calc['Denom'] = df_calc['GPUs'] + df_calc['Cb']
    df_calc['Denom'] = df_calc['Denom'].replace(0, np.nan)
    df_calc['VPT'] = df_calc['TokenCost'] * df_calc['I'] / df_calc['Denom']
    df_calc['TFR'] = df_calc['VPT'] / df_calc['TokenCost']

    df_calc['VPT'] = df_calc['VPT'].fillna(0)
    df_calc['TFR'] = df_calc['TFR'].fillna(0)

    return df_calc

# ------------------------------------------------------------
# Create time series data
# ------------------------------------------------------------
df_base = pd.DataFrame.from_dict(scenarios, orient='index').reset_index().rename(columns={'index':'Entity'})

# Generate parameter values for animation
param_values = np.linspace(PARAM_MIN, PARAM_MAX, N_FRAMES)

# Create dataframe for animation
data_frames = []
for param_val in param_values:
    df_t = calculate_metrics(df_base, param_val)
    df_t['param_value'] = param_val
    data_frames.append(df_t)

df_animation = pd.concat(data_frames, ignore_index=True)

print(f"Animation will show {PARAM_TO_VARY} varying from {PARAM_MIN:,} to {PARAM_MAX:,}")
print(f"Total frames: {N_FRAMES}")

# ------------------------------------------------------------
# Create Economic VPT Scatter Animation
# ------------------------------------------------------------
scale = 1e6

# Calculate global min/max for consistent axes
x_min = df_animation['I'].min() / scale
x_max = df_animation['I'].max() / scale
y_min = df_animation['VPT'].min()
y_max = df_animation['VPT'].max()

# Add margins
x_margin = (x_max - x_min) * 0.1
y_margin = (y_max - y_min) * 0.1

fig, ax = plt.subplots(figsize=(12, 8))

# Color scheme for entities
colors = plt.cm.plasma(np.linspace(0, 1, len(df_base)))
entity_colors = dict(zip(df_base['Entity'], colors))

# Create a placeholder colorbar that will be updated
cbar = None

def animate(frame):
    ax.clear()

    # Get data for current frame
    param_current = param_values[frame]
    df_frame = data_frames[frame].copy()

    # Calculate positions
    x = df_frame['I'] / scale
    y = df_frame['VPT']

    # Size based on TFR (scaled)
    sizes = (df_frame['TFR'] / df_frame['TFR'].max()) * 800 + 50

    # Color based on TFR value
    colors_frame = df_frame['TFR']

    # Create scatter plot
    sc = ax.scatter(x, y, s=sizes, c=colors_frame, cmap='plasma',
                   edgecolors='k', alpha=0.85, linewidths=0.8)

    # Add labels for all entities
    for idx, row in df_frame.iterrows():
        x_pos = row['I'] / scale
        y_pos = row['VPT']

        # Determine label position
        if x_pos > (x_min + x_max) / 2:
            ha = 'right'
            offset_x = -8
        else:
            ha = 'left'
            offset_x = 8

        ax.annotate(row['Entity'], (x_pos, y_pos),
                   xytext=(offset_x, 5), textcoords='offset points',
                   fontsize=7, ha=ha, va='bottom',
                   bbox=dict(boxstyle='round,pad=0.3', facecolor='white',
                            edgecolor='gray', alpha=0.8, linewidth=0.5),
                   arrowprops=dict(arrowstyle='-', color='gray', lw=0.5, alpha=0.6))

    # Styling
    ax.set_xlabel(f'Interface Intelligence (×{int(scale):,})', weight='bold', fontsize=12)
    ax.set_ylabel('Value Per Token (VPT) $', weight='bold', fontsize=12)

    # Format title based on parameter
    if PARAM_TO_VARY == "LLM_score":
        display_val = f"{param_current*100:.1f}%"
    else:
        display_val = f"{param_current:,.0f}"

    ax.set_title(f'Economic Efficiency: VPT vs Interface Intelligence(II)\n{PARAM_TO_VARY} = {display_val}',
                weight='bold', fontsize=14, pad=15)

    # Set consistent axis limits
    ax.set_xlim(x_min - x_margin, x_max + x_margin)
    ax.set_ylim(y_min - y_margin, y_max + y_margin)

    ax.grid(linestyle=':', alpha=0.4)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

    # Add colorbar only on first frame
    global cbar
    if frame == 0 and cbar is None:
        cbar = plt.colorbar(sc, ax=ax, pad=0.02)
        cbar.set_label('Token Financial Ratio (TFR)', weight='bold', fontsize=11)

    plt.tight_layout()

# Create animation
print("Creating VPT scatter animation... This may take 1-2 minutes.")
anim = FuncAnimation(fig, animate, frames=N_FRAMES, interval=50, repeat=True)

# Save as GIF
gif_path = os.path.join(BASE_DIR, f"CII_VPT_Scatter_{PARAM_TO_VARY}.gif")
writer = PillowWriter(fps=20)
anim.save(gif_path, writer=writer)
print(f"✅ Animation saved: {gif_path}")

# Display in notebook
try:
    from IPython.display import Image
    display(Image(gif_path))
except:
    print("Animation created. Open the GIF file to view.")

plt.close()

# ------------------------------------------------------------
# Create comparison: Start vs End scatter plots
# ------------------------------------------------------------
fig_comp, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 7))

# Format display values
if PARAM_TO_VARY == "LLM_score":
    display_min = f"{PARAM_MIN*100:.1f}%"
    display_max = f"{PARAM_MAX*100:.1f}%"
else:
    display_min = f"{PARAM_MIN:,.0f}"
    display_max = f"{PARAM_MAX:,.0f}"

# Start frame
df_start = calculate_metrics(df_base, PARAM_MIN)
x_start = df_start['I'] / scale
y_start = df_start['VPT']
sizes_start = (df_start['TFR'] / df_start['TFR'].max()) * 800 + 50

sc1 = ax1.scatter(x_start, y_start, s=sizes_start, c=df_start['TFR'],
                  cmap='plasma', edgecolors='k', alpha=0.85, linewidths=0.8)

# Add annotations for all entities
for idx, row in df_start.iterrows():
    x_pos = row['I'] / scale
    y_pos = row['VPT']

    if x_pos > x_start.median():
        ha = 'right'
        offset_x = -8
    else:
        ha = 'left'
        offset_x = 8

    ax1.annotate(row['Entity'], (x_pos, y_pos),
               xytext=(offset_x, 5), textcoords='offset points',
               fontsize=6, ha=ha, va='bottom',
               bbox=dict(boxstyle='round,pad=0.3', facecolor='white',
                        edgecolor='gray', alpha=0.8, linewidth=0.5),
               arrowprops=dict(arrowstyle='-', color='gray', lw=0.5, alpha=0.6))

ax1.set_xlabel(f'II (×{int(scale):,})', weight='bold')
ax1.set_ylabel('VPT ($)', weight='bold')
ax1.set_title(f'START: {PARAM_TO_VARY} = {display_min}', weight='bold', fontsize=13)
ax1.grid(linestyle=':', alpha=0.4)
cbar1 = plt.colorbar(sc1, ax=ax1)
cbar1.set_label('TFR', weight='bold')

# End frame
df_end = calculate_metrics(df_base, PARAM_MAX)
x_end = df_end['I'] / scale
y_end = df_end['VPT']
sizes_end = (df_end['TFR'] / df_end['TFR'].max()) * 800 + 50

sc2 = ax2.scatter(x_end, y_end, s=sizes_end, c=df_end['TFR'],
                  cmap='plasma', edgecolors='k', alpha=0.85, linewidths=0.8)

# Add annotations for all entities
for idx, row in df_end.iterrows():
    x_pos = row['I'] / scale
    y_pos = row['VPT']

    if x_pos > x_end.median():
        ha = 'right'
        offset_x = -8
    else:
        ha = 'left'
        offset_x = 8

    ax2.annotate(row['Entity'], (x_pos, y_pos),
               xytext=(offset_x, 5), textcoords='offset points',
               fontsize=6, ha=ha, va='bottom',
               bbox=dict(boxstyle='round,pad=0.3', facecolor='white',
                        edgecolor='gray', alpha=0.8, linewidth=0.5),
               arrowprops=dict(arrowstyle='-', color='gray', lw=0.5, alpha=0.6))

ax2.set_xlabel(f'II (×{int(scale):,})', weight='bold')
ax2.set_ylabel('VPT ($)', weight='bold')
ax2.set_title(f'END: {PARAM_TO_VARY} = {display_max}', weight='bold', fontsize=13)
ax2.grid(linestyle=':', alpha=0.4)
cbar2 = plt.colorbar(sc2, ax=ax2)
cbar2.set_label('TFR', weight='bold')

plt.suptitle('CII Economic VPT Comparison: Start vs End', weight='bold', fontsize=16, y=0.98)
plt.tight_layout()
comp_path = os.path.join(BASE_DIR, f"CII_VPT_Comparison_{PARAM_TO_VARY}.png")
plt.savefig(comp_path, bbox_inches='tight', dpi=200)
print(f"✅ Comparison saved: {comp_path}")
plt.show()
plt.close()

print(f"\n✅ VPT scatter animation complete!")
print(f"   Parameter varied: {PARAM_TO_VARY} from {PARAM_MIN:,} to {PARAM_MAX:,}")
print(f"   Files saved to: {BASE_DIR}")
```
